```python
import pandas as pd
df = pd.read_csv("../data/Sample - Superstore.csv",encoding="latin1")
```

```python
#stage 4 : Exploratory Data Analysis

# 1 : Identify top performing product categories
df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
```

```python
# 2 : Analyze sales distribution across regions
df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
```

```python
# 3 : Determine which customer segments generate the most revenue
df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
```

```python
# 4 : Analyze monthly sales trend
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Month"] = df["Order Date"].dt.month
df.groupby("Month")["Sales"].sum()
```

```python
# 4 : Analyze yearly sales trend
df["Year"] = df["Order Date"].dt.year
df.groupby("Year")["Sales"].sum()
```
